```python
encoding = tiktoken.get_encoding("gpt2")
ints = encoding.encode("Akwirw ier")
print(ints)
```

```python
word = encoding.decode(ints)
print(word)
```

```python
for i in ints:
    print(f"{i} -> {encoding.decode([i])}")
```
